I selected the wave IX of the Arab Barometer data. https://www.arabbarometer.org/survey-data/data-downloads/

Arab Barometer is a nonpartisan research network that provides insight into the social, political, and economic attitudes and values of ordinary citizens across the Arab world. They have been conducting public opinion surveys in the Middle East and North Africa since 2006. 

```python
# I am using pandas to read in the data
import pandas as pd  # importing pandas library for data manipulation and analysis.

data = pd.read_csv(
    "ArabBarometer_WaveIX_English_Dataset_NOV_2026.csv"
)  # This line reads the csv file into a dataFrame called data.
```

```python
## Preliminary data exploration
```

```python
data  # checking that the the dataframe has between one thousand and onne million rows

data.head()  # seeing the first 5 rows of the DataFrame to get an overview of the data
```

```python
# Here, I am picking one quantitative column to analyze
# I picked column "Q1001" which asks about the respondent's age.
# This information was found from the codebook provided with the dataset.
# This column is quantitative because it contains numerical values representing the age of the respondents

data[
    "Q1001"
].describe()  # using the describe() method to get summary statistics of the "Q1001" column
```

```python
## Preliminary column-of-interest cleaning
```

```python
# remove NA values from the "Q1001" column which is coded as 99999 when respondents refused to answer
data_clean = data[data["Q1001"] != 99999]

data_clean[
    "Q1001"
].describe()  # using the describe() method to get summary statistics of the cleaned "Q1001" column
```

```python
## Mean, median, and mode of the cleaned "Q1001" column using pandas methods
```

```python
mean_age = data_clean[
    "Q1001"
].mean()  # calculating the mean age of respondents in the cleaned "Q1001" column

mean_age
# Mean age is 40.712250712250714
```

```python
median_age = data_clean[
    "Q1001"
].median()  # calculating the median age of respondents in the cleaned "Q1001" column

median_age
# Median age is 39.0
```

```python
mode_age = data_clean[
    "Q1001"
].mode()  # calculating the mode age of respondents in the cleaned "Q1001" column

mode_age
# Mode age is 40.0
```
