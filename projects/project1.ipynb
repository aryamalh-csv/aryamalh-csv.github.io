## The Hard Way - Computing mean, median, and mode of the cleaned age column without using pandas methods

```python
# cleaning this up a bit
import csv  # importing the csv library

df = "ArabBarometer_WaveIX_English_Dataset_NOV_2026.csv"
column = "Q1001"  # respondent age variable
min_age = 18  # valid range from the codebook
max_age = 120

# Read the column from the CSV, keeping only valid ages
ages_list = []  # initializing an empty list to hold valid ages
rows_read = 0  # counter for the number of rows read from the CSV file
with open(df, newline="") as f:
    for row in csv.DictReader(
        f
    ):  # iterating through each row in the CSV file using a DictReader to read the data as dictionaries
        rows_read += 1  # incrementing the counter for each row read
        age = float(
            row[column]
        )  # converting the age value from the current row to a float
        if min_age <= age <= max_age:  # checking if the age is within the valid
            ages_list.append(age)

n = len(ages_list)  # this now contains the number of valid ages in the ages_list
print(rows_read)  # 11264
print(n)  # 11232

# 32 no response responses were dropped from the dataset.
```

```python
# mean
total = 0  # initializing a variable to hold the total sum of ages
for age in ages_list:  # for loop to iterate through each age in the ages_list
    total += age
mean = total / n

print(mean)  # printing the mean age of respondents in the cleaned "Q1001" column
```

```python
# median
ordered = sorted(ages_list)
mid = n // 2
if n % 2 == 1:
    median = ordered[mid]
else:
    median = (ordered[mid - 1] + ordered[mid]) / 2

print(median)  # printing the median age of respondents in the cleaned "Q1001" column
```

```python
# mode
# count each value in a dictionary, then find the most common

counts = {}
for age in ages_list:
    if age in counts:
        counts[age] += 1
    else:
        counts[age] = 1

highest = 0
for c in counts.values():
    highest = max(highest, c)

mode = None
for age, c in counts.items():
    if c == highest:
        mode = age
        break

print(mode)
```

```python
# create visualization

# Filter out None and NaN values (and keep ages within 18 to 120)
clean_ages = [a for a in ages_list if a is not None and a == a and 18 <= a <= 120]

start_age = 18  # starting age for the bins
end_age = 120  # ending age for the bins
bin_width = 10

# Group ages into bins (handles both integers and decimals)
bins = []
for start in range(
    start_age, end_age + 1, bin_width
):  # iterating through the range of ages from start_age to end_age in increments of bin_width
    end = min(
        start + bin_width - 1, end_age
    )  # calculating the end of the current bin, ensuring it does not exceed end_age
    if (
        end == end_age
    ):  # Including the upper boundary (<= end_age) on the final bin so age 120 is counted
        count = sum(1 for a in clean_ages if start <= a <= end_age)
    else:
        count = sum(1 for a in clean_ages if start <= a < start + bin_width)
    bins.append(
        (f"{start}-{end}", count)
    )  # appending a tuple containing the bin label and the count of ages in that bin to the bins list

# Print the chart
max_count = max(
    (c for _, c in bins), default=1
)  # finding the maximum count of ages in any bin to determine the scaling factor for the bar chart
scale = (
    35 / max_count if max_count > 0 else 1
)  # calculating the scaling factor for the bar chart to ensure that the longest bar fits within 35 characters

print(
    f"Age Distribution (n = {len(clean_ages)})"
)  # printing the title of the chart along with the total number of valid ages in the cleaned "Q1001" column
print("-" * 45)
for (
    label,
    count,
) in (
    bins
):  # iterating through each bin label and count in the bins list to print the bar chart
    bar = "🩷" * int(count * scale)
    print(f"{label:<8} | {bar} ({count})")
```
